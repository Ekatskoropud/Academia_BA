### Ejercicio 1 BMI calculator

```python
def calculator_BMI():
    while True: 
        weight = input('Enter your weight in kilograms ')
        if weight.isdigit():
            x = int(weight)
            break
        else: 
            print ('Enter your weight in kilograms using only numbers ')        
    while True: 
        height = input('Enter your height in centimeters ')
        if height.isdigit():
            y = int(height) ** 2
            break
        else: 
            print ('Enter your height in kilograms using only numbers ')        
    BMI = round((x / y) * 10000)
    if BMI <= 16:
        message = f"Your BMI is {BMI} - Significant underweight"
    elif BMI >= 17 and BMI <= 18:
        message = f"Your BMI is {BMI} - Insufficient (deficiency) body weight" 
    elif BMI >=19 and BMI <= 25:
        message = f"Your BMI is {BMI} - Normal body weight"
    elif BMI >= 26 and BMI <= 30:
        message = f"Your BMI is {BMI} - Excess body weight (pre-obesity)"
    elif BMI > 30:
        message = f"Your BMI is {BMI} - You may be obese. If you are not a professional athlete, we recommend that you consult your doctor."
    
    return message 
```

```python
calculator_BMI()
```

### Ejercicio 2 Temperature converter


```python
def temperature_converter():
    conversion = {
        'C': { 
            'K': lambda c:(c + 273.15),               #Celsius to Kelvin
            'F': lambda c:((c * 9/5) + 32)            #Celsius to Fahrenheit
        },          
        'K': {
            'C': lambda k:(k - 273.15),               #Kelvin to Celsius
            'F': lambda k:((k - 273.15) * 9/5 + 32)   #Kelvin to Fahrenheit
        },
        'F': {
            'C': lambda f:((f - 32) * 5/9),            #Fahrenheit to Celsius 
            'K': lambda f:((f - 32) * 5/9 + 273.15)    #Fahrenheit to Kelvin
        }
    }

    while True: 
        user_input = input('Enter temperature with unit (e.g. 25C, 298K, 77F) ').strip().upper()
        if user_input[-1] in ["C","F","K"]:
            try:
                x = float(user_input[:-1])
                unit = user_input[-1]
                break
            except ValueError: 
                print('Temperature value must be number')
        else:
            print ('Enter your temperature using only numbers and unit C, K, F  ')      

    # Creation new dictionary for the results of multiplicaciones 
    results = {} 
    for target_unit, formula in conversion[unit].items():
        results[target_unit] = formula(x)

    messages = []
    for target_unit, value in results.items():
        messages.append(f"{x}{unit} = {value:.2f}{target_unit}")

    return " ".join(messages)
```

```python
temperature_converter()
```

### Ejercicio 3 Words counter

```python
import re
from collections import Counter

def words_calculator():
        while True: 
            txt = input('Enter your text: ')
            if not txt.strip():
                print('Text cannot be empty. Enter your text.')
            elif txt.isascii():
                txt = txt.lower()
                words = re.findall(r"\b\w+\b", txt) 
                break
            else:
                print('Text contains invalid characters. Re-enter with the correct values.')
            
        words_count = Counter(words)
        for word, count in words_count.items():
            print(word, "-", count)
        
        # Average word length
        total_chars = sum(len(word) for word in words)
        average = round(total_chars / len(words))
        print('Average length of the words in your text is', average)
```

### Ejercicio 4 Reversed dictionary

```python
user_dictionary = {'a': 1, 'b': 2, 'c': 3, 'd':1}
user_dictionary
```

```python
def reverse_dictionary(dictionaryname):

    reversed_dict = {v: k for k, v in dictionaryname.items()}
    
    return reversed_dict
```

```python
reverse_dictionary(user_dictionary)
```

```python
from collections import defaultdict

def dublicate_reverse_dictionary(dictionaryname):
    reversed_dict = defaultdict(list)
    for key, value in dictionaryname.items():
        reversed_dict[value].append(key)
    
    return dict(reversed_dict)
```

```python
dublicate_reverse_dictionary(user_dictionary)
```

### Ejercicio 1 Imported document words' counter

```python
import re
from collections import defaultdict

def words_counter(filename):
    with open(filename, 'r', encoding='utf-8') as file:
        txt = file.read()
        
    wordlist = re.findall(r"\b\w+\b", txt)
    normalized_words = []
    for word in wordlist:
        clean_word = word.lower()
        if clean_word:
            normalized_words.append(clean_word)
        
    counter = defaultdict(int)
    for word in normalized_words:
        counter[word] += 1
        
    sorted_words = sorted(counter.items())
    
    grouped = defaultdict(dict)
    for word, counter in sorted_words:
        first_letter = word[0].lower()   
        grouped[first_letter][word] = counter
    return dict(grouped)
```

### Ejercicio 2 Data type interpreter

```python
data = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]
```

```python
import re

#[+-]? Finding a number with a plus or minus sign in front of it
# \d+ finding a number with more than one digit
#(\.\d+)? finding a number with a fractional part separated by a period
#([eE][+-]?\d+)? numbers with an exponent

pattern = re.compile(r'^[+-]?\d+(\.\d+)?([eE][+-]?\d+)?$') 

def extract_numbers(user_data):
    numbers = []
    not_numbers = []

    for item in user_data:
        if isinstance(item, (int, float)):  
            numbers.append(item)
        elif isinstance(item, str):          
            if pattern.match(item):
                numbers.append(float(item))
            else:
                not_numbers.append(item)
        elif isinstance(item, (list, tuple)):
            sub_numbers, sub_not_numbers = extract_numbers(item)
            numbers.extend(sub_numbers)
            not_numbers.extend(sub_not_numbers)      

    return numbers, not_numbers
```

### Ejercicio 1 Password generator

```python
import random 
     
def create_password(length = 16, uppercase_letter = True, lowercase_letter = True, number = True, symbol = False):
    lowercase_letters = 'abcdefghijklmnopqrstuvwxyz'
    numbers = '1234567890'
    uppercase_letters = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
    symbols = "!#$%&*+-.=?@"

    min_password = ""
    if uppercase_letter: 
        min_password += random.choice(uppercase_letters)
    if lowercase_letter:
        min_password += random.choice(lowercase_letters)
    if number:
        min_password += random.choice(numbers)
    if symbol: 
        min_password += random.choice(symbols) 
    
    if not min_password: 
        raise ValueError ('The characters from which the password will be generated are not selected. Select at least one character type.')

    if len(min_password) > length: 
        raise ValueError ('the length cannot be less than the number of characters selected for generation')   

    characterSet = ""
    if uppercase_letter: 
        characterSet += uppercase_letters
    if lowercase_letter: 
        characterSet += lowercase_letters
    if number:
        characterSet += numbers
    if symbol: 
        characterSet += symbols
    
    reminder_length = length - len(min_password)
    remainder = "".join(random.choices(characterSet, k=reminder_length))
    
    password = min_password + remainder
    password = "".join(random.sample(password, len(password)))
    
    return password
```

### Ejercicio 2 Football match statistics

```python
from collections import defaultdict 

def footbal_stat(filename):
    stats = defaultdict(lambda: {"win": 0, "draw": 0, "lose": 0, "goals": 0, "missedGoals": 0, "rating": 0})
    
    with open(filename, 'r', encoding='utf-8') as file:
        for line in file: 
            parts = line.strip().split()
            clubs = []
            scores = []
             
            current_club = []
            for word in parts:
                if word.replace('-', '').isdigit():
                    if current_club:
                        clubs.append(" ".join(current_club))
                        current_club = []
                    scores.extend(word.split('-'))
                else:
                    current_club.append(word)
            
            if current_club:
                clubs.append(" ".join(current_club))

            if len(scores) == 2 and len(clubs) == 2: 
                score1, score2 = int(scores[0]), int(scores[1])
                team1, team2 = clubs[0], clubs[1]

                stats[team1]["goals"] += score1
                stats[team2]["goals"] += score2
                stats[team1]["missedGoals"] += score2
                stats[team2]["missedGoals"] += score1

                if score1 > score2:
                    stats[team1]["win"] += 1
                    stats[team1]["rating"] += 3
                    stats[team2]["lose"] += 1
                elif score1 < score2:
                    stats[team2]["win"] += 1
                    stats[team2]["rating"] += 3
                    stats[team1]["lose"] += 1
                else:
                    stats[team1]["draw"] += 1
                    stats[team2]["draw"] += 1
                    stats[team1]["rating"] += 1
                    stats[team2]["rating"] += 1
    return dict(stats)
```
