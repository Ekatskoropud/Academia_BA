#### Ejercicio 1-2

```python
import requests
import pandas as pd
```

```python
BASE_URL = "https://jsonplaceholder.typicode.com/"
```

```python
posts_response = requests.get(f"{BASE_URL}/posts")
users_response = requests.get(f"{BASE_URL}/users")
todos_response = requests.get(f"{BASE_URL}/todos")

print(f"Posts: {len(posts_response.json())} records, status code {posts_response.status_code}")
print(f"Users: {len(users_response.json())} records, status code {users_response.status_code}")
print(f"Tasks: {len(todos_response.json())} records, status code {todos_response.status_code}\n")
```

#### Ejercicio 3

```python
invalid_response = requests.get(f"{BASE_URL}/posts/999999")
print(f"Status code: {invalid_response.status_code}\n")
```

#### Ejercicio 4

```python
new_post = {
    "title": "New post",
    "body": "My best new post!",
    "userId": 1
}
post_response = requests.post(f"{BASE_URL}/posts", json=new_post)

print("Response:", post_response.json())
print(f"Status code: {post_response.status_code}\n")
```

#### Ejercicio 5

```python
updated_data = {"title": "Updated title"}
patch_response = requests.patch(f"{BASE_URL}/posts/101", json=updated_data)

print("Response:", patch_response.json())
print(f"Status code: {patch_response.status_code}\n")
```

#### Ejercicio 6

```python
delete_response = requests.delete(f"{BASE_URL}/posts/101")

print("Response:", delete_response.json())
print(f"Status code: {delete_response.status_code}\n")
```
